## Confirm search is semantic, not keyword-only

```python
semantic_query = "AI systems that answer patient questions online"
results2 = semantic_search.search_papers(semantic_query, top_k=5)
for r in results2:
    print(f"{r['similarity_score']:.3f}  {r['title']}")
```
