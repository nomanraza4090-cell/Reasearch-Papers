# 08 — Research Gap Analysis

```python
import sys, json
sys.path.insert(0, "..")
from src import config, semantic_search, research_gap

topic = "large language models in clinical decision support"
papers = semantic_search.search_papers(topic, top_k=20)
print(f"{len(papers)} papers retrieved for gap analysis")
```

```python
gaps = research_gap.analyze_research_gaps(papers)
print(json.dumps(gaps, indent=2, default=str)[:3000])
```

## Save report

```python
report_path = config.REPORTS_DIR / "research_gap_report.json"
with open(report_path, "w") as f:
    json.dump({"topic": topic, **gaps}, f, indent=2, default=str)
print("Saved to", report_path)
```
